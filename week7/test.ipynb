```r
# Cài đặt thư viện
install.packages("ggplot2")

library(ggplot2)

# Miền giá trị
data <- data.frame(x = c(-4, 4))

# Vẽ biểu đồ
ggplot(data, aes(x = x)) +
  stat_function(
    fun = dnorm,
    aes(color = "Z (Standard Normal)"),
    linewidth = 1.2
  ) +
  stat_function(
    fun = dt,
    args = list(df = 5),
    aes(color = "T (df = 5)"),
    linewidth = 1.2
  ) +
  stat_function(
    fun = dt,
    args = list(df = 30),
    aes(color = "T (df = 30)"),
    linewidth = 1.2
  ) +
  scale_color_manual(
    values = c(
      "Z (Standard Normal)" = "blue",
      "T (df = 5)" = "red",
      "T (df = 30)" = "green"
    )
  ) +
  labs(
    title = "Comparison of Z and T Distributions",
    x = "Value",
    y = "Probability Density",
    color = "Distribution"
  ) +
  theme_minimal()
```

```r
# 1. Tạo 1 triệu số ngẫu nhiên phân phối đều U(0,1)
set.seed(123)

N <- 1000000
x <- runif(N, min = 0, max = 1)

# 2. Tính độ lệch chuẩn
SD <- sd(x)

# 3. Chia thành 10.000 mẫu, mỗi mẫu có 100 số
n <- 100
B <- N / n

samples <- matrix(x, nrow = n, ncol = B)

# Tính trung bình của từng mẫu
sample_means <- colMeans(samples)

# 4. Tính sai số chuẩn
SE <- SD / sqrt(n)

# Sai số chuẩn nếu sử dụng toàn bộ 1 triệu số
SE_1M <- SD / sqrt(N)

# 5. In kết quả
cat("Trung bình dữ liệu:", mean(x), "\n")
cat("Độ lệch chuẩn SD:", SD, "\n")
cat("Sai số chuẩn SE (n=100):", SE, "\n")
cat("SD của các trung bình mẫu:", sd(sample_means), "\n")
cat("SE của trung bình 1 triệu số:", SE_1M, "\n")

# 6. Vẽ đồ thị
par(mfrow = c(1, 2))

# Đồ thị 1: Dữ liệu ban đầu
hist(x,
     breaks = 50,
     probability = TRUE,
     col = "skyblue",
     main = "Do lech chuan (SD)",
     xlab = "Gia tri",
     ylab = "Mat do")

# Đồ thị 2: Trung bình của các mẫu
hist(sample_means,
     breaks = 40,
     probability = TRUE,
     col = "orange",
     main = "Sai so chuan (SE)",
     xlab = "Trung binh mau",
     ylab = "Mat do")

# Đường cong chuẩn tham chiếu
curve(
  dnorm(x, mean = mean(sample_means), sd = SE),
  add = TRUE,
  col = "red",
  lwd = 2
)
```
