```r

# Download dataset from Kaggle
url <- "https://www.kaggle.com/api/v1/datasets/download/hopesb/student-depression-dataset"

zip_path <- "/content/student-depression.zip"
download.file(url, zip_path, mode = "wb", method = "libcurl")

# Extract dataset
data_dir <- "/content/student-depression"
dir.create(data_dir, showWarnings = FALSE, recursive = TRUE)

unzip(zip_path, exdir = data_dir)

# Find CSV file
csv_path <- list.files(
  data_dir,
  pattern = "\\.csv$",
  recursive = TRUE,
  full.names = TRUE,
  ignore.case = TRUE
)

print(csv_path)

# Import CSV
student_data <- read.csv(
  csv_path[1],
  header = TRUE,
  stringsAsFactors = FALSE,
  check.names = TRUE
)

# Check data
head(student_data)
```

```r
summary(student_data)
barplot(table(student_data$Gender), main = "Gender Distribution", col = c("lightblue", "pink"), xlab = "Gender", ylab = "Count")
```
