```r

url <- "https://www.kaggle.com/api/v1/datasets/download/stackoverflow/so-survey-2017"

zip_path <- "/content/so-survey-2017.zip"

download.file(
  url,
  destfile = zip_path,
  mode = "wb",
  method = "libcurl"
)

data_dir <- "/content/so-survey-2017"

dir.create(data_dir, showWarnings = FALSE, recursive = TRUE)

file_list <- unzip(zip_path, list = TRUE)

print(file_list$Name)

unzip(zip_path, exdir = data_dir)

csv_path <- list.files(
  data_dir,
  pattern = "survey_results_public\\.csv$",
  recursive = TRUE,
  full.names = TRUE,
  ignore.case = TRUE
)

if (length(csv_path) == 0) {
  stop("Cannot find survey_results_public.csv")
}

survey <- read.csv(
  csv_path[1],
  header = TRUE,
  stringsAsFactors = FALSE,
  check.names = FALSE
)

cat("Import successful!\n")
cat("Rows:", nrow(survey), "\n")
cat("Columns:", ncol(survey), "\n")

head(survey)
```
